```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    'Date': ['Nov 1', 'Nov 2', 'Nov 3', 'Nov 4', 'Nov 5'],
    'Visitors': [100, 150, 120, 200, 180],
    'Internship': [15, 20, 18, 30, 25],
    'Training': [10, 15, 12, 20, 18],
    'Service_Requests': [5, 8, 6, 12, 10]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Check missing values
print(df.isnull().sum())

# Remove duplicate records
df = df.drop_duplicates()

# Fill missing numerical values
df = df.fillna(df.mean(numeric_only=True))

print("Cleaned Data:")
print(df)
```

```python
print("Dataset Shape:", df.shape)

print("Column Names:")
print(df.columns)

print("Statistical Summary:")
print(df.describe())
```

```python
total = df['Internship'].sum()

average = df['Internship'].mean()

maximum = df['Internship'].max()

minimum = df['Internship'].min()

print("Total Applications:", total)
print("Average Applications:", average)
print("Maximum Applications:", maximum)
print("Minimum Applications:", minimum)
```

```python
print("Total Training Enquiries:",
      df['Training'].sum())

print("Average Training Enquiries:",
      df['Training'].mean())
```

```python
print("Total Service Requests:",
      df['Service_Requests'].sum())

print("Average Service Requests:",
      df['Service_Requests'].mean())
```

```python
activities = {
    'Internship': df['Internship'].sum(),
    'Training': df['Training'].sum(),
    'Service Requests': df['Service_Requests'].sum()
}

plt.bar(activities.keys(), activities.values())

plt.xlabel("Business Activities")
plt.ylabel("Total")
plt.title("Business Activities Comparison")

plt.show()
```

```python
marketing = pd.DataFrame({
    'Channel': ['Google', 'Instagram', 'Facebook',
                'Google', 'Instagram'],
    'Visitors': [100, 150, 80, 200, 120],
    'Applications': [15, 25, 10, 30, 20]
})

print(marketing)
```

```python
result = marketing.groupby('Channel')['Visitors'].sum()

print(result)

result.plot(kind='bar')

plt.title("Marketing Channel Analysis")
plt.xlabel("Marketing Channel")
plt.ylabel("Total Visitors")
plt.show()
```

```python
customers = pd.DataFrame({
    'Customer_ID': [101, 102, 103, 104, 105],
    'Service': ['Web', 'App', 'Web', 'IT', 'App'],
    'Requests': [5, 8, 3, 10, 6]
})

print(customers)

print("Total Customers:",
      customers['Customer_ID'].count())

print("Total Requests:",
      customers['Requests'].sum())

print(customers.groupby('Service')['Requests'].sum())
```

```python
service = customers.groupby('Service')['Requests'].sum()

plt.pie(service, labels=service.index,
        autopct='%1.1f%%')

plt.title("Service Requests Distribution")

plt.show()
```

```python
from sklearn.linear_model import LinearRegression

X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)

y = np.array([100, 150, 120, 200, 180])

model = LinearRegression()

model.fit(X, y)

future = np.array([[6]])

prediction = model.predict(future)

print("Predicted Website Visitors:",
      prediction[0])
```

```python
plt.scatter(X, y, label="Actual Visitors")

plt.plot(X, model.predict(X),
         label="Predicted Trend")

plt.xlabel("Days")
plt.ylabel("Visitors")

plt.title("Website Visitors Forecasting")

plt.legend()

plt.show()
```
